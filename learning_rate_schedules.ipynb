**STEP DECAY**

new_learning_rate = initial_lr * drop_factor ** (np.floor(epoch/drop_every))

```python
import numpy as np

def step_decay(initial_lr, epoch, drop_factor, drop_every):
	LR = initial_lr*drop_factor**(np.floor(epoch/drop_every))
	return LR
```

**EXPONENTIAL DECAY**

new_learning_rate = initial_lr * np.exp(-decay_rate/epoch)

```python
def exponential_decay(initial_lr, epoch, decay_rate):
	LR = initial_lr* np.exp(-decay_rate*epoch)
	return LR
```

**POLYNOMIAL DECAY**

new_learning_rate = (initial_lr - final_lr)*(1-epoch/total_epochs)**power + final_lr

```python
def polynomial_decay(initial_lr, final_lr, epoch, total_epochs, power):
	epoch = min(epoch, total_epochs)
	LR = (initial_lr - final_lr)*(1-epoch/total_epochs)**power + final_lr

	return np.round(LR, 4)
```
