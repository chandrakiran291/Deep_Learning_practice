```python
import numpy as np 

def sigmoid(x):
    return 1/(1+np.exp(-x))

def tanh(x):
    return np.tanh(x)

def relu(x):
    return np.maximum(0, x)

def leaky_relu(x, alpha=0.01):
    return np.where(x>0, x, alpha*x)

def softmax(x):
    exp_x = np.exp(x, np.max(x, axis=0, keepdims=True))
    return exp_x/np.sum(exp_x, axis=1, keepdims=True)
```

```python
X = np.linspace(-10, 10, 500)

y_sigmoid = sigmoid(X)
y_tanh = tanh(X)
y_relu = relu(X)
y_leaky_relu = leaky_relu(X)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 10))
plt.subplot(2, 2, 1)
plt.plot(X, y_sigmoid)
plt.title("sigmoid activation function")
plt.grid(True)
plt.xlabel("x")
plt.ylabel("sigmoid(x)")

plt.subplot(2,2,2)
plt.plot(X, y_tanh)
plt.title("tanh activation function")
plt.grid(True)
plt.xlabel("x")
plt.ylabel("tanh(x)")

plt.subplot(2,2,3)
plt.plot(X, y_relu)
plt.title("relu activation function")
plt.grid(True)
plt.xlabel("x")
plt.ylabel("relu(x)")  

plt.subplot(2,2,4)
plt.plot(X, y_leaky_relu)
plt.title("leaky relu activation function")
plt.grid(True)
plt.xlabel("x")
plt.ylabel("leaky relu(x)")
```

```python
#softmax activation function

class1 =[]
class2 =[]
class3 =[]

for i in X:
    probabilities = softmax(np.array([[i, 1, 0]]))
    class1.append(probabilities[0, 0])
    class2.append(probabilities[0, 1])
    class3.append(probabilities[0, 2])  

plt.plot(X, class1, label="class 1")
plt.plot(X, class2, label="class 2")
plt.plot(X, class3, label="class 3")
plt.title("softmax activation function")
plt.grid(True)
plt.xlabel("x")
plt.ylabel("probability")
plt.legend()
```
