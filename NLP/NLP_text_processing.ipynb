### Required steps to follow:

Step 1 - Lowercasing  
Step 2 - Tokenization  
Step 3 - Smart Stop Word Removal (preserve sentiment words like `not`, `never`)  
Step 4 - Lemmatization



### You can start writing your code below

```python
lowered_reviews = []
for sen in reviews:
  sen = sen.lower()
  lowered_reviews.append(sen)

lowered_reviews
```

```python
tokenized_reviews = [word_tokenize(review) for review in lowered_reviews]
tokenized_reviews
```

```python
stop_words = set(stopwords.words('english'))
sentiment_words = {'no', 'not', 'never'}
stop_words = stop_words - sentiment_words
filtered_reviews = [[word for word in review if word.isalpha() and word.lower() not in stop_words] 
                    for review in tokenized_reviews]

filtered_reviews
```

```python
lemmatizer = WordNetLemmatizer()
lemmatized_reviews = []
for each in filtered_reviews:
  lemmatized_reviews.append([lemmatizer.lemmatize(w) for w in each])

lemmatized_reviews
```
